# 简单的线性回归模型练习

数据矩阵：

$$
X=
\begin{bmatrix}
x_{11}&x_{12}\\
x_{21}&x_{22}
\end{bmatrix}
$$

通过最小二乘法求解：

$$
w=(X^TX)^{-1}X^Ty
$$

得到预测：

$$
\hat y=Xw
$$

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data =  pd.read_excel("LDH_ML_training_data.xlsx",sheet_name="LDH_data")
x = data.iloc[:,[1,3,4,5]].values #不包括末尾索引值
y = data.iloc[:,6].values
mean = x.mean(axis = 0)
std = x.std(axis = 0)
x_scaled = (x-mean)/std
w = np.linalg.inv(x_scaled.T@x_scaled
                  )@x_scaled.T@y
b = y.mean() - x_scaled.mean(axis=0) @ w
y_pre =x_scaled@w + b
print(w)
mse = np.mean(y-y_pre)**2
print(mse)
```

```python
plt.scatter(y, y_pre,color = "red")

plt.xlabel("Real")
plt.ylabel("Predict")

plt.plot(
    [y.min(), y.max()],
    [y.min(), y.max()]
)

plt.show()
```
